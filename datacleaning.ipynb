```python
import pandas as pd
import numpy as np

data = {
    "ID": [101, 102, 103, 103, 104, 105, 106, 107, 108],
    "Name": ["Alice", "Bob", "Charlie", "Charlie", "David", None, "Emma", "Ryan", "Sophia"],
    "Age": [25, np.nan, 30, 30, 28, 22, 26, np.nan, 29],
    "Salary": [50000, 60000, None, None, 55000, 45000, 48000, 52000, None],
    "Joining_Date": [
        "2024-01-10",
        "12/02/2024",
        "2024/03/15",
        "2024/03/15",
        "15-04-2024",
        None,
        "2024-05-10",
        "2024/06/11",
        "11-07-2024"
    ]
}

df = pd.DataFrame(data)

print("------ RAW DATASET ------")
print(df)

print("\n------ MISSING VALUES ------")
print(df.isnull().sum())

df = df.drop_duplicates()

df["Name"] = df["Name"].fillna("Unknown")
df["Age"] = df["Age"].fillna(26)
df["Salary"] = df["Salary"].fillna(52000)

df["Joining_Date"] = pd.to_datetime(df["Joining_Date"], errors='coerce')

df["Joining_Date"] = df["Joining_Date"].fillna(pd.Timestamp("2024-01-01"))

print("\n------ CLEANED DATASET ------")
print(df)

print("\n------ FINAL CHECK ------")
print("Duplicate IDs:", df["ID"].duplicated().sum())
print("Incorrect Dates:", df["Joining_Date"].isnull().sum())

df.to_csv("cleaned_dataset.csv", index=False)

print("\nDataset cleaned successfully!")

```
